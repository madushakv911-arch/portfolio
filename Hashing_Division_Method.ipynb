```python
%%writefile q7_hashing.c

#include <stdio.h>

#define TABLE_SIZE 10
#define N 8

int hashFunction(int key)
{
    return key % TABLE_SIZE;
}

void insert(int hashTable[], int key)
{
    int index = hashFunction(key);
    int originalIndex = index;

    printf("\nInserting %d\n", key);
    printf("Hash index = %d\n", index);

    while (hashTable[index] != -1)
    {
        printf("Collision at index %d\n", index);

        index = (index + 1) % TABLE_SIZE;

        if (index == originalIndex)
        {
            printf("Hash table is full.\n");
            return;
        }
    }

    hashTable[index] = key;

    printf("Inserted %d at index %d\n", key, index);
}

void displayTable(int hashTable[])
{
    printf("\n-----------------------------\n");
    printf("       HASH TABLE\n");
    printf("-----------------------------\n");

    printf("Index\tValue\n");

    for (int i = 0; i < TABLE_SIZE; i++)
    {
        if (hashTable[i] == -1)
            printf("%d\tEMPTY\n", i);
        else
            printf("%d\t%d\n", i, hashTable[i]);
    }

    printf("-----------------------------\n");
}

int hashSearch(int hashTable[], int key, int *comparisons)
{
    int index = hashFunction(key);
    int start = index;

    *comparisons = 0;

    while (hashTable[index] != -1)
    {
        (*comparisons)++;

        if (hashTable[index] == key)
            return index;

        index = (index + 1) % TABLE_SIZE;

        if (index == start)
            break;
    }

    return -1;
}

int linearSearch(int arr[], int n, int key, int *comparisons)
{
    *comparisons = 0;

    for (int i = 0; i < n; i++)
    {
        (*comparisons)++;

        if (arr[i] == key)
            return i;
    }

    return -1;
}

int main()
{
    int songIDs[N] = {
        105, 210, 315, 420,
        525, 630, 735, 840
    };

    int searchKeys[] = {
        105, 525, 840, 999
    };

    int numSearchKeys = 4;

    int hashTable[TABLE_SIZE];

    for (int i = 0; i < TABLE_SIZE; i++)
    {
        hashTable[i] = -1;
    }

    printf("DATA STRUCTURES AND ALGORITHMS - ASSIGNMENT 2\n");
    printf("Question 7 - Hashing using Division Method\n");
    printf("Roll No: 34\n");

    printf("\nSong IDs:\n");

    for (int i = 0; i < N; i++)
    {
        printf("%d ", songIDs[i]);
    }

    printf("\n");

    printf("\n========== INSERTION ==========\n");

    for (int i = 0; i < N; i++)
    {
        insert(hashTable, songIDs[i]);
        displayTable(hashTable);
    }

    printf("\n========== FINAL HASH TABLE ==========\n");
    displayTable(hashTable);

    printf("\n========== SEARCH COMPARISON ==========\n");

    for (int i = 0; i < numSearchKeys; i++)
    {
        int key = searchKeys[i];

        int hashComparisons;
        int linearComparisons;

        int hashResult =
            hashSearch(hashTable, key, &hashComparisons);

        int linearResult =
            linearSearch(songIDs, N, key, &linearComparisons);

        printf("\nSearching for %d\n", key);

        if (hashResult != -1)
            printf("Hashing: Found at index %d, Comparisons = %d\n",
                   hashResult, hashComparisons);
        else
            printf("Hashing: Not Found, Comparisons = %d\n",
                   hashComparisons);

        if (linearResult != -1)
            printf("Linear Search: Found at array index %d, Comparisons = %d\n",
                   linearResult, linearComparisons);
        else
            printf("Linear Search: Not Found, Comparisons = %d\n",
                   linearComparisons);
    }

    float loadFactor = (float)N / TABLE_SIZE;

    printf("\n========== ANALYSIS ==========\n");

    printf("Number of elements = %d\n", N);
    printf("Hash table size = %d\n", TABLE_SIZE);
    printf("Load factor = %d/%d = %.2f\n",
           N, TABLE_SIZE, loadFactor);

    printf("\nHash Function: h(k) = k %% %d\n", TABLE_SIZE);

    printf("\nAverage Hashing Search Complexity: O(1)\n");
    printf("Worst-case Hashing Search Complexity: O(n)\n");
    printf("Linear Search Complexity: O(n)\n");

    printf("\nSpace Complexity of Hash Table: O(m)\n");

    return 0;
}
```
